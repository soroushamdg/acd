```python
from selenium.webdriver import Chrome
from selenium.webdriver.chrome.options import Options
import random
from time import sleep

opts = Options()
opts.add_argument("--mute-audio")

av_engine = Chrome(options=opts, executable_path=r'C:\Users\Soroush\PycharmProjects\ac_downloader\github\release\engine_database\acd_sengine.acde')
```

```python

av_engine.get('https://www.aparat.com/iranpythoneers')
all_vids = av_engine.find_elements_by_class_name('thumbnail-video')
rndm_vids = random.choices(all_vids,k=3)
vids_links = [vid.find_element_by_tag_name('a').get_attribute('href') for vid in rndm_vids]
```

```python
for vid in vids_links:
    try:
        av_engine.set_network_conditions(offline=False,latency=0,download_throughput=240000,upload_throughput=240000)  # maximal throughput
        av_engine.get(vid)
    except:
        continue
    else:
        av_engine.set_network_conditions(offline=False,latency=0,download_throughput=5000,upload_throughput=5000)  # maximal throughput
    sleep(300)
```
