```python

def get_size_format(n, suffix="B"):
     # converts bytes to scaled format (e.g KB, MB, etc.)
     for unit in ["", "K", "M", "G", "T", "P"]:
         if n < 1024:
             return f"{n:.2f} {unit}{suffix}"
         n /= 1024
```

```python
## preparing requests
s = requests.Session()
for cookie in course_cookies:
    s.cookies.set(name=cookie["name"],
                  value=cookie["value"],
                  domain=  cookie.get("domain"),
 path=  cookie.get("path"),
secure=  cookie.get("secure"),
    )
```

```python
data = s.get(download_url)
with open('test.zip','wb') as f:
    f.write(data.content)
```

```python
with s.get(download_url,stream=True) as r:
    print(r.headers)
    full_size =int(r.headers['Content-length'])
    f = open('filename3.zip',"wb")
    print(f.closed)
    dl_size=0
    for chunk in r.iter_content(chunk_size=1024):
        f.write(chunk)
        dl_size+=1024
        clear_output(wait=True)
        print(f'downloaded {get_size_format(dl_size)}/{get_size_format(full_size)}')
    f.close()
    print(f.closed)
```
