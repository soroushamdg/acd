```python
def get_size_format(n, suffix="B"):
     # converts bytes to scaled format (e.g KB, MB, etc.)
     for unit in ["", "K", "M", "G", "T", "P"]:
         if n < 1024:
             return f"{n:.2f}{unit}{suffix}"
         n /= 1024
```

```python
global pgbar

def ftp_list_directory(ftp : FTP):
    for item in ftp.nlst():
         current = ftp.pwd()
         try:
             ftp.cwd(item)
         except:
             print(item, 'FILE', get_size_format(ftp.size(item)))
         else:
             ftp.cwd(current)
             print(item, 'DIR')

def ftp_cw_directory(ftp : FTP, foldername : str):
    try:
        ftp.cwd(foldername)
    except Exception as msg:
        print(msg)
        return False
    else:
        return True


def ftp_generate_file_handler(file):

    def file_writer_handler(data):
        file.write(data)
        print(f'Downloaded : {len(data)}')

    return file_writer_handler

def ftp_download_file(ftp : FTP, path : str):
    filename = path.split('/')[-1]

    file = open(filename,'wb')

    ftp.retrbinary("RETR "+path,file.write)
    file.close()

```
