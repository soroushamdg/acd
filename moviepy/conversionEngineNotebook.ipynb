VIDEO AND AUDIO

```python
streamIndex = streamParser.parseindexstream(r'C:\Users\Soroush\PycharmProjects\acdii\test_adobeconnects\multiple_audio_video_course\indexstream.xml')
main_path = r'C:\Users\Soroush\PycharmProjects\acdii\test_adobeconnects\multiple_audio_video_course'
print(streamIndex['video'])
```

```python
# conneting audios
audios = []
for item in streamIndex['audio']:
    try:
        audios.append((AudioFileClip(str(pathlib.Path(main_path).joinpath(item.filename))).set_start(tuple([milisec2sec(item.delay)]))))
    except Exception as msg:
        print(msg)
        continue
mixed_audio=CompositeAudioClip(audios)
mixed_audio.fps=audios[0].fps

# connecting videos
videos=[]
for item in streamIndex['video']:
    try:
        video = VideoFileClip(str(pathlib.Path(main_path).joinpath(item.filename)))
        video = video.fx(vfx.resize,width=960)
        videos.append(video.set_start(tuple([milisec2sec(item.delay)])))
    except Exception as msg:
        print(msg)
        continue
mixed_video = CompositeVideoClip(videos)

mixed_video.audio = mixed_audio

# export whole items
mixed_video.write_videofile('multiple_audio_videos.mp4',
                            threads=100,
                            # fps=24,
                            # bitrate="800K",
                            # verbose=False,
                            # audio_bitrate="64K",
                            preset="ultrafast"
                            )
```
