```python
%load_ext autotime

import pathlib
from PIL import Image
from moviepy.editor import *
from xmlparser import streamParser
import settings
```

```python
def milisec2sec(miliseconds):
    return (miliseconds/1000)

def generate_blank_screen_image(width,height,output=None):
    background = Image.new(mode="RGB", size=(width,height), color=(0,0,0))
    info = Image.open(r'C:\Users\Soroush\PycharmProjects\acdii\src\novideo.jpg')
    offset = int((width/2)-300), int((height/2)-300)
    background.paste(info, offset)
    background.save(output if output else settings.return_temp_path(filename=str(width)+str(height) + '.jpg'))
    return output if output else settings.return_temp_path(filename=str(width)+str(height) + '.jpg')


def create_blank_screen_video(duration, size=(1920, 1088), output=None):
    img = generate_blank_screen_image(size[0],size[1],output=output)
    clip = ImageClip(img).set_duration(duration)
    clip = concatenate_videoclips([clip],method="compose")
    clip.write_videofile(output if output else settings.return_temp_path(filename=str(duration) + '.mp4'), fps=25, verbose=False,
        logger=None)
    return output if output else settings.return_temp_path(filename=str(duration) + '.mp4')


def generate_blank_screen_image(width,height,output=None):
    background = Image.new(mode="RGB", size=(width,height), color=(0,0,0))
    info = Image.open(settings.src_path('novideo.jpg'))
    offset = int((width/2)-300), int((height/2)-300)
    background.paste(info, offset)
    background.save(output if output else settings.return_temp_path(filename=str(width)+str(height) + '.jpg'), format='JPEG', optimize=True, quality= 20 if width * height >= 422500 else 80)
    return output if output else settings.return_temp_path(filename=str(width)+str(height) + '.jpg')


def create_blank_screen_video(duration, size, output=None):
    img =  generate_blank_screen_image(size[0],size[1],output=output)
    clip = ImageClip(img).set_duration(duration)
    clip = concatenate_videoclips([clip],method="compose")
    clip.write_videofile(output if output else settings.return_temp_path(filename=str(duration) + f'.{settings.return_output_format()}'), fps=1,verbose=True,
        logger='bar')
    clip.close()
    return output if output else settings.return_temp_path(filename=str(duration) + f'.{settings.return_output_format()}')
```

# CONCAT VIDEOS AND BLANK SCREENS


```python
streamIndex = streamParser.parseindexstream(r'C:\Users\Soroush\PycharmProjects\acdii\test_adobeconnects\multiple_audio_video_course\indexstream.xml')
main_path = r'C:\Users\Soroush\PycharmProjects\acdii\test_adobeconnects\multiple_audio_video_course'
print(streamIndex['video'])
```

```python
videos=[]
for item in streamIndex['video']:
    try:
        video = VideoFileClip(str(pathlib.Path(main_path).joinpath(item.filename)))
        videos.append(video.set_start(tuple([milisec2sec(item.delay)])))
    except Exception as msg:
        print(msg)
        continue
for index, item in enumerate(videos):
    if index ==0:
        if item.start != 0:
            blank_screen = VideoFileClip(create_blank_screen_video(duration=item.start,size=(item.w,item.h)))
            videos.insert(0,blank_screen)
            continue
    else:
        if item.start - videos[index-1].start - videos[index-1].duration > 0 :
            blank_screen = VideoFileClip(create_blank_screen_video(duration= item.start - (videos[index-1].start + videos[index-1].duration)))
            videos.insert(index,blank_screen)
print(videos)
```

```python
from pyffmpeg import FFmpeg
import pyffmpeg
pyffmpeg.FFmpeg
import time

ff = FFmpeg(r'C:\Users\Soroush\PycharmProjects\acdii\test_adobeconnects\outputs')
ff.overwrite = True
ff.loglevel = ff.loglevels[0] # quiet log level
audio_command = r"-i C:\Users\Soroush\PycharmProjects\acdii\test_adobeconnects\outputs\multiple_audio_video_course_audio.mp3"
for bitrate, format in [("90k","mkv"),("60k","mkv"),("30k","mkv")]:
    start = time.time()
    args = (audio_command,
    " ".join([f"-i {video.filename}" for video in videos]),
      "-filter_complex",
      f'"{" ".join([f"[{index+1 if audio_command else index}:v]fps=25[v{index+1 if audio_command else index}];" if item.fps != 25 else "" for index, item in enumerate(videos)]).lstrip()}',
       f'{" ".join([f"[v{index+1 if audio_command else index}]" if item.fps != 25 else f"[{index+1 if audio_command else index}:v]" for index, item in enumerate(videos)]).rstrip()} concat=n={len(videos)}:v=1{":a=0" if audio_command else ""} [outv]"',
      "-map [outv]",
        f"-map 0:a:0" if audio_command else "",
        # '-c:v libx264 -c:a aac',
        f'-b:v {bitrate} -threads 3 -preset ultrafast',
        # "-movflags +faststart",
        '-stats',
      f'multiple_audio_video_course_video{bitrate}.{format}')
    result = ff.options(" ".join(args))
    end = time.time()
    print(bitrate,format,end-start)

```
